```python
print(df['Unit_Price'].describe())
```

```python
Q1 = df['Unit_Price'].quantile(0.25)
Q3 = df['Unit_Price'].quantile(0.75)
IQR = Q3 - Q1
```

```python
Lower = Q1 - 1.5 * IQR
Higher = Q3 + 1.5 * IQR
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", Lower)
print("Higher Bound:", Higher)
```

```python
outliers = df[
    (df['Unit_Price']<Lower) | (df['Unit_Price']>Higher)
]
print("Number of Outliers: ", len(outliers))
```

```python
print(outliers[['Order_ID', 'Product', 'Category', 'Quantity', 'Unit_Price', 'Sales_Amount']])
```

```python
df['Product']=df['Product'].str.strip().str.title()
```

```python
sus = df[
    (df['Product']=='Laptop') &
    (df['Unit_Price']>100000)
]
print(sus)
```

```python
df['City']=df['City'].fillna(df['City'].mode()[0])
df['Customer_Type']=df['Customer_Type'].fillna(df['Customer_Type'].mode()[0])
df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
df['Discount_Pct']=df['Discount_Pct'].fillna(df['Discount_Pct'].mean())
df['Order_Date']=df['Order_Date'].fillna(df['Order_Date'].mode()[0])
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'], errors='coerce')
df['Discount_Pct']=pd.to_numeric(df['Discount_Pct'], errors='coerce')
```

***Exploratory Data Analysis***


```python
print("Total Order:", df['Order_ID'].nunique())
print("Total Customers:",df['Customer_ID'].nunique())
print("Total Sales:", df['Sales_Amount'].sum())
print("Total Profit:", df['Profit'].sum())
print("Total Quantity:", df['Quantity'].sum())
print("Average Order Value:", df['Sales_Amount'].mean())
```

```python
print(df[['Quantity', 'Unit_Price', 'Discount_Pct', 'Sales_Amount', 'Profit']].describe())
```

```python
category_analysis = df.groupby('Category').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(category_analysis)
```

```python
region_analysis=df.groupby('Region').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(region_analysis)
```

```python
city_analysis=df.groupby('City').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)
print(city_analysis)
```

```python
print(city_analysis.head(10))
```

```python
sales_channel_analysis=df.groupby('Sales_Channel').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(sales_channel_analysis)
```

```python
customer_type_analysis=df.groupby('Customer_Type').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(customer_type_analysis)
```

```python
product_analysis=df.groupby('Product').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(product_analysis)
```

```python
discount_analysis=df.groupby('Discount_Pct').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_values('Sales', ascending=False)

print(discount_analysis)
```

```python
trend_analysis=df.groupby('Order_Date').agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
).sort_index()

print(trend_analysis)
```

```python
monthly_trend_analysis=df.groupby(df['Order_Date'].dt.to_period('M')).agg(
    Sales=('Sales_Amount', 'sum'),
    Profit=('Profit', 'sum'),
    Quantity=('Quantity', 'sum'),
    Orders=('Order_ID','nunique')
)
print(monthly_trend_analysis)
```
